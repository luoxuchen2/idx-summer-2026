Keeps more rows in the dataset.
Useful when a column is important but has some missing values.
Median imputation works well for skewed data like LivingArea, LotSizeSquareFeet, or TaxAnnualAmount.
Allows the model to still use partially complete records.

Cons:

You are filling in values that were not actually observed.
Mean/median imputation can reduce natural variation in the data.
If too much data is missing, imputation may become misleading.
Simple imputation does not always capture the true relationship between variables.

```python
target = 'ClosePrice'

numeric_cols = [
    'LivingArea',
    'DaysOnMarket',
    'LotSizeSquareFeet',
    'YearBuilt',
    'BathroomsTotalInteger',
    'BedroomsTotal',
    'GarageSpaces',
    'Latitude',
    'Longitude',
    'FireplacesTotal',
    'TaxAnnualAmount',
    'AssociationFee',
    'Stories',
    'MainLevelBedrooms'
]

categorical_cols = [
    'City',
    'CountyOrParish',
    'MLSAreaMajor',
    'SubdivisionName',
    'HighSchoolDistrict',
    'ElementarySchoolDistrict',
    'Levels',
    'PostalCode'
]

binary_cols = [
    'PoolPrivateYN',
    'FireplaceYN',
    'NewConstructionYN',
    'AttachedGarageYN',
    'ViewYN',
    'WaterfrontYN',
    'BasementYN'
]

# Keep only columns that exist in the dataframe
numeric_cols = [col for col in numeric_cols if col in df.columns]
categorical_cols = [col for col in categorical_cols if col in df.columns]
binary_cols = [col for col in binary_cols if col in df.columns]

keep_cols = [target, 'source_month'] + numeric_cols + categorical_cols + binary_cols
df = df[keep_cols].copy()

df.head()
```

```python
# remove extreme outliers 

outlier_cols = [target] + numeric_cols

for col in outlier_cols:
    upper = df[col].quantile(0.99)
    df_model = df[df[col].isna() | (df[col] <= upper)]

df.shape
```

```python
# Handle missing numeric values with flags + median imputation

for col in numeric_cols:
    df_model[col + '_missing_flag'] = df_model[col].isna().astype(int)
    df_model[col] = df_model[col].fillna(df_model[col].median())

#This lets you keep the row, but also preserves whether the value was originally missing.    
```

```python
#Encode binary Y/N columns

for col in binary_cols:
    df_model[col] = df_model[col].map({
        'Y': 1,
        'N': 0,
        True: 1,
        False: 0,
        'True': 1,
        'False': 0,
        'true': 1,
        'false': 0
    })

    df_model[col + '_missing_flag'] = df_model[col].isna().astype(int)
    df_model[col] = df_model[col].fillna(0)
```

```python
# Handle categorical missing values
for col in categorical_cols:
    df_model[col] = df_model[col].fillna('Unknown').astype(str)

#To avoid creating too many dummy columns, group rare values as "Other":

for col in categorical_cols:
    top_values = df_model[col].value_counts().head(25).index
    df_model[col] = df_model[col].where(df_model[col].isin(top_values), 'Other')
```

```python
# One-hot encode categorical columns
df_encoded = pd.get_dummies(
    df_model,
    columns=categorical_cols,
    drop_first=True
)

df_encoded.head()
```

```python
# Normalize numeric features

from sklearn.preprocessing import StandardScaler


scale_cols = numeric_cols + [col + '_missing_flag' for col in numeric_cols]

scale_cols = [col for col in scale_cols if col in df_encoded.columns]

scaler = StandardScaler()
df_encoded[scale_cols] = scaler.fit_transform(df_encoded[scale_cols])
```

```python
# Train/test split by most recent month
df_encoded['source_month'] = df_encoded['source_month'].astype(float).astype(int).astype(str)

months = sorted(df_encoded['source_month'].unique())

test_month = months[-1]

print("Available months:", months)
print("Test month:", test_month)
```

```python
X_window = 6

train_months = months[-(X_window + 1):-1]

train_df = df_encoded[df_encoded['source_month'].isin(train_months)].copy()
test_df = df_encoded[df_encoded['source_month'] == test_month].copy()

print("Training months:", train_months)
print("Testing month:", test_month)
print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

drop_cols = [target, 'source_month']
feature_cols = [col for col in df_encoded.columns if col not in drop_cols]

results = []

for X_window in [3, 6, 9, 12]:
    train_months = months[-(X_window + 1):-1]
    test_month = months[-1]

    train_df = df_encoded[df_encoded['source_month'].isin(train_months)].copy()
    test_df = df_encoded[df_encoded['source_month'] == test_month].copy()

    X_train = train_df[feature_cols]
    y_train = train_df[target]

    X_test = test_df[feature_cols]
    y_test = test_df[target]

    model = LinearRegression()
    model.fit(X_train, y_train)

    preds = model.predict(X_test)

    mae = mean_absolute_error(y_test, preds)
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    r2 = r2_score(y_test, preds)

    results.append({
        'training_window_months': X_window,
        'train_months': ', '.join(train_months),
        'test_month': test_month,
        'mae': mae,
        'rmse': rmse,
        'r2': r2
    })

results_df = pd.DataFrame(results)
results_df.sort_values('mae')
```

```python
output_file = folder / "residential_cleaned_preprocessed.csv"

df_encoded.to_csv(output_file, index=False)

print("Saved cleaned CSV to:", output_file)
```
