# Week 1 Deliverables
I was able to access the files with FileZilla
- ListingKey: the primary key and uniquely identifies records in the Property resource
- Living area (decimal) - the total amount of area that is livable in the property
- BedroomTotal (int): the total amount of bedroom in the property
- BathroomFull (int): the the area that has a toliet, sink, bathrub and showerhead, all four itmes 
- close price (decimal) : the amount of money paid by the buyer to the seller
- Days On the Market (int): Shows how long the property stayed on the market. Longer time may suggest weaker demand.
- PropertyType:	General property category; should be Residential
- PropertySubType:	More specific property type; should be SingleFamilyResidence
- City:	City of the listing address
- PostalCode: ZIP code of the property
- LotSizeArea::	Larger lots usually affect property value.
- LotSizeUnits:	Needed to understand whether lot size is in square feet, acres, etc.
- YearBuilt:	Older vs. newer homes may have different values.
- CountyOrParish:	Helps group properties by county.
- Latitude and Longitude:	Useful for geographic modeling or future school district joins.
- GarageSpaces:	Parking can affect value.
- ParkingTotal:	Total parking availability may matter, especially by location.
- PoolPrivateYN:	A pool may affect price in California markets.
- ViewYN:	Views can increase property value.
- WaterfrontYN:	Waterfront status can strongly affect price.
- FireplaceYN:	Amenity that may influence value.
- AssociationFee:	HOA cost can affect affordability and buyer demand.
- AssociationYN:	Indicates whether the property has an HOA.
